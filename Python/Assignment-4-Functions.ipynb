```python
def OddEven(num):
    if(num%2 == 0):
        res = f"{num} is Even number"
    else:
        res = f"{num} is Odd number"
    return res

input_num = int(input("Enter a number: "))
res = OddEven(input_num)
print(res)
    
```

```python
def Eligible(gender, age):
    
    if(gender == "Male"):
        marriage_age = 24
    else:
        marriage_age = 21
    
    if(age >= marriage_age):
        out = "ELIGIBLE"
    else:
        out = "NOT ELIGIBLE"
    return out

gender = input("Your Gender: ")
age = int(input("Your Age: "))
result = Eligible(gender,age)
print(result)
        
```

```python
def Percentage(tot,numOfsubs):
    avg = tot/numOfsubs
    return avg

tot = 0
sub_1 = int(input("Subject1= "))
sub_2 = int(input("Subject2= "))
sub_3 = int(input("Subject3= "))
sub_4 = int(input("Subject4= "))
sub_5 = int(input("Subject5= "))

lists = [sub_1,sub_2,sub_3,sub_4,sub_5]
for lt in lists:
    tot += lt
result = Percentage(tot,len(lists))
print("Total: ", tot)
print("Percentage: ", result)
```
