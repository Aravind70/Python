```python
#Scenario - 1
def AmountCategory(amount):
    res = ''
    if(amount > 0):
        res = "Positive"
    elif(amount < 0):
        res = "Negative"
    else:
        res = "Zero"
    return res

amount = int(input("Enter the amount: "))
result = AmountCategory(amount)
print(f"Amount is {result}")
```

```python
#Scenario - 2
def SumPassCode(lst):
    sum = 0
    for num in lst:
        sum += int(num)
    return sum

passCode = input("Enter the Passcode: ")
lst = list(passCode)
sum = SumPassCode(lst)
print(sum)
```

```python
#Scenario - 3
#transc_id = "IDI1234DIT"

def ReverseTranID(transc_id):
    rev = ''
    for id in range(len(transc_id), 0, -1):
        rev += transc_id[id-1]
    return rev

transc_id = input("Enter the transaction ID: ")
result = ReverseTranID(transc_id)    
print(result)
    
```

```python
#Scenario - 4
def PrimeNumber(user_id):
    count = 0
    for num in range(user_id):
        if(num+1 == 1):
            count += 1
            continue
        if(user_id % (num+1) == 0):
            count += 1
    return count

user_id = int(input("Enter the ID: "))
result = PrimeNumber(user_id)
if(result == 2):
    print("Prime Number")
else:
    print("Not a Prime Number")
```

```python
#Scenario - 5
def Fact(num, res):
    if num == 0:
        return res
    res = res * num
    num = num-1
    return Fact(num, res)

fact_num = int(input("Enter the number: "))
result = Fact(fact_num,1)
print(f"Factorial number is {result}")

```

```python
#Scenario - 6
def ArmstrongNum(lst,len_armNum):
    total = 0
    for num in lst:
        value = int(num) ** len_armNum
        total += value
    return total

arm_num = input("Enter the number: ")
lst = list(arm_num)
len_armNum = len(arm_num)
armst_num = ArmstrongNum(lst,len_armNum)
if(armst_num == int(arm_num)):
    print("Armstrong number")
else:
    print("Not Armstrong number")
```

```python
#Scenario - 7

password = "art"

def PasswordChange(password,first_char, last_char):
    lst = list(password)
    lst[0] = first_char
    lst[len(password) - 1] = last_char
    return lst

#input
password = input("Enter the password: ")
first_char = input("Enter the character to swap with first character: ")
last_char = input("Enter the character to swap with last character: ")
result = PasswordChange(password, first_char, last_char)

strPassword = ""
for char in result:
    strPassword += char
print("Password is %s" % strPassword)

```

```python
#scenario - 8

number = int(input("Enter the decimal number: "))
binary_num = ''

while number > 0:
    rem = number % 2
    number = number // 2    
    binary_num += str(rem)
print(binary_num[::-1])
print("".join(reversed(binary_num)))
    
```

```python
#Scenario - 9
def FindLongestWord(lst):
    res = 0
    for word in lst:
        max_char = len(word)
        if max_char > res:
            res = max_char
            
    return word

sentence = input("Write a sentence: ")
lst = sentence.split()
result = FindLongestWord(lst)
print("Longest word in sentence is", result)
```

```python
#Scenario - 10
def IsAnagrams(word_1,word_2):
    
    word1_len = len(word_1)
    word1_count = 0
    word2_count = 0
    dummy_string = ""
    count = 0
    
    for ch_1 in word_1:
        
        if ch_1 in dummy_string:
            count += 1
            continue
        
        if ch_1 in word_2:
            word1_count = word_1.count(ch_1)
            word2_count = word_2.count(ch_1)
            dummy_string += ch_1
            if word1_count == word2_count:
                count += 1
    if(word1_len == count):
        return 0
    else:
        return 1

            


#input
word_1 = input("Enter the first word: ")
word_2 = input("Enter the second word: ")
result = IsAnagrams(word_1, word_2)
if(result == 0):
    print("Given words are Anagrams")
else:
    print("Given words are not Anagrams")
```
