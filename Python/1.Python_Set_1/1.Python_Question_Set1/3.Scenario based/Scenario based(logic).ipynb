```python
#Scenario - 1
def EligibleForVote(age):
    if(age >= 18):
        return "Eligible"
    else:
        return "Not Eligible"
age = int(input("Enter the age: "))
result = EligibleForVote(age)
print(result)
```

```python
#Scenario - 2

def FindMaxNumber(lst):
    max = 0
    for num in lst:
        if max < num:
            max = num
    return max

numbers = input("Enter the inputs by comma seperated(1,2,3): ")
lst = []
for num in numbers:
    if(num != ","):
        lst.append(int(num))
max_num = FindMaxNumber(lst)
print(max_num)
    
```

```python
#Scenario -3
def FindBonusAmount(salary):
    if(salary > 50000):
        bonus = int(salary * (10 / 100))
    else:
        bonus = "No bonus"
    return bonus
        
salary = int(input("Enter the salary amount: "))
result = FindBonusAmount(salary)
print(result)
```

```python
#Scenario - 4
def OddEven(number):
    res = ""
    if(number%2 == 0):
        res = "Even"
    else:
        res = "Odd"
    return res

num = int(input("Enter the number: "))
res = OddEven(num)
print(f"Given numder is {res}")
```

```python
#Scenario - 5
def ReverseString(word):
    rev = ''
    for num in range(len(word),0,-1):
        rev += word[num-1]
    return rev

word = input("Type the word: ")
res = ReverseString(word)
print("Reverse string is ", res)
        
        
```

```python
#Scenario- 6
def GradingSystem(mark):
    res = ''
    if(mark > 40):
        res = "PASSED"
    else:
        res = "FAILED"
    return res


mark = int(input("Enter the mark of the subject: "))
res = GradingSystem(mark)
print("Student is ", res)
```

```python
#Scenario - 7
def FindDiscount(price):
    if(price > 100):
        final_price = int(price - (price * (20 / 100)))
    else:
        final_price = "No Discount"
    return final_price
        
price = int(input("Enter the price: "))
result = FindDiscount(price)
print(result)
```

```python
#Scenario - 9
def LeapYear(year):
    res = ''
    if(year%4 == 0 and year%100!=0):
        res = 'Leap Year'
    elif(year%400 == 0):
        res = 'Leap Year'
    else:
        res = 'Not Leap Year'
    return res

year = int(input("Enter the year: "))
result = LeapYear(year)
print(result)
```

```python
#Scenario - 10
def EvenNumbers(lst):
    new_lst = []
    for num in lst:
        if(num%2 == 0):
            new_lst.append(num)
    return new_lst
    

numbers = input("Enter the numbers(comma seperated): ")
lst = []
for num in numbers:
    if(num != ","):
        lst.append(int(num))
evenNumList = EvenNumbers(lst)
print(evenNumList)
```
