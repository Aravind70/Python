2.Write a Python program to print the first 5 even numbers. Expected Output:
2 4 6 8 10



```python
i = 0
even_num = 1
while i < 5:
    if even_num%2 == 0:
        i += 1
        print(even_num)
    even_num += 1
```

3.Write a Python program to print the sum of numbers from 1 to 10. 
Expected Output:55



```python
i = 1
sum = 0
while i <= 10:
    sum += i
    i += 1
print(sum)
```

4.Write a Python function to find the factorial of 5. Expected Output:
120



```python
fact_num = 5
mul = 1
for itr in range(fact_num,1,-1):
    mul = mul * itr
print(mul)
```

5.Write a Python program to check if a number is prime. Expected Output (for input 7):
7 is a prime number



```python
prime_num = 7

count = 0
for num in range(prime_num):
    if((num+1) == 1):
        count += 1
        continue
    if(prime_num % (num+1) == 0):
        count += 1
        
if(count == 2):
    print(f"{prime_num} is Prime Number")
else:
    print(f"{prime_num} is not Prime Number")
```
